```python
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
# See For Plotly
```

```python
df = pd.read_csv('retail_sales.csv')
df.head()
```

1 - Summary Statistics

```python
df.describe()
```

2 - Detect Outliers

```python
# Age Column
sns.boxplot(x ='age', data=df)
```

```python
# Quantity Column
sns.boxplot(x='quantity', data=df)
```

```python
# Unit Price Column
sns.boxplot(x='unit_price', data=df)
```

```python
# Revenue Column
sns.boxplot(x='revenue', data=df)
```

3 - Histogram

```python
# Revenue Histgram
sns.histplot(x=df['revenue'], kde=True, bins=10)
```

```python
# Unit_Price Histogram
sns.histplot(x=df['unit_price'], kde=True, bins=10)
```

```python
# Quantity Histogram
sns.histplot(x=df['quantity'], kde=True, bins=10)
```

```python
# Age Histogram
sns.histplot(x=df['age'], kde=True, bins=10)
```

4 - Check Correlation

```python
df_copy = df.copy()
df_copy.drop(columns=['transaction_id',  'date', 'customer_id', 'gender', 'product_category'], inplace=True)
sns.heatmap(df_copy.corr(), annot=True, cmap='RdBu')
```
